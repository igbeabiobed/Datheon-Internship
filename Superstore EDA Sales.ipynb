# Data wrangling and cleaning
### Standardize column names

```python
df.columns = (df.columns.str.strip().str.lower().str.replace(' ', '_'))

print(df.columns.tolist())
```

### Check for duplicate records

```python
print("Total duplicate records:", df.duplicated().sum())
```

### Remove duplicates:

```python
df = df.drop_duplicates()

print("Duplicates removed successfully!")
```

### Profile missing values

```python
missing_values = pd.DataFrame({'Missing_Count': df.isnull().sum(), 'Missing_Percentage': (df.isnull().mean() * 100).round(2)})
missing_values = missing_values.sort_values(
    by='Missing_Percentage',
    ascending=False)

display(missing_values)
```

### Visualize missing values

```python
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values Heatmap")
plt.show()
```

### Handle missing values

```python
categorical_columns = df.select_dtypes(include='object').columns
for col in categorical_columns: df[col] = df[col].fillna('Unknown')
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
df['sales'] = pd.to_numeric(df['sales'], errors='coerce')
df = df.dropna(subset=['order_date', 'sales'])
```

```python
print (df)
```

### Convert order dates into datetime format

```python
df['order_date'] = pd.to_datetime(df['order_date'])
```

```python
df['order_date'].dtype
```

### Extract year, month and year-month

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['year_month'] = (df['order_date'].dt.to_period('M'))
```

```python
df[['order_date', 'year', 'month', 'year_month']].head()
```

# Exploratory Data Analysis
### Analyze customer purchase volume

```python
unique_customers = df['customer_id'].nunique()
print("Total Unique Customers:", unique_customers)
```

```python
total_orders = df['order_id'].nunique()
print("Total Orders:", total_orders)
```

```python
total_sales = df['sales'].sum()
print("Total Sales:", round(total_sales, 2))
```

```python
average_order_value = (df.groupby('order_id')['sales'].sum().mean())

print("Average Order Value:", round(average_order_value, 2))
```

### Analyze top customers

```python
customer_sales = (df.groupby('customer_id')['sales'].sum().sort_values(ascending=False))

customer_sales.head(10)
```

### Visualize the top 10 customers:

```python
plt.figure(figsize=(12, 6))

customer_sales.head(10).plot(kind='bar',color='steelblue')
plt.title("Top 10 Customers by Sales")
plt.xlabel("Customer ID")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.show()
```

### Analyze monthly sales seasonality

```python
monthly_sales = (df.groupby('year_month')['sales'].sum().reset_index())

monthly_sales.head()
```

```python
print(monthly_sales.head())
print(monthly_sales.dtypes)
print(type(monthly_sales['year_month'].iloc[0]))
```

```python
monthly_sales = (df.groupby(df['order_date'].dt.to_period('M') )['sales'].sum().reset_index())

monthly_sales.columns = ['year_month','sales']
```

```python
monthly_sales['year_month'] = (monthly_sales['year_month'].dt.to_timestamp())
```

### Create a line chart:

```python
plt.figure(figsize=(15, 6))
sns.lineplot(data=monthly_sales, x='year_month', y='sales', marker='o')
plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Year-Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Identify the best-performing month

```python
best_month = monthly_sales.loc[monthly_sales['sales'].idxmax()]

print("Highest Sales Month:")
print(best_month)
```

### Identify the lowest-performing month:

```python
worst_month = monthly_sales.loc[monthly_sales['sales'].idxmin()]

print("Lowest Sales Month:")
print(worst_month)
```

### Analyze category-wise revenue
Determine which product categories generate the highest revenue.

```python
category_sales = (df.groupby('category')['sales'].sum().sort_values(ascending=False))

print(category_sales)
```

### Visualize the results:

```python
plt.figure(figsize=(10, 6))
category_sales.plot( kind='bar',color='teal')
plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

### Analyze category-wise profit

```python
category_profit = (df.groupby('category')['profit'].sum().sort_values(ascending=False))

print(category_profit)
```

```python
plt.figure(figsize=(10, 6))

category_profit.plot(kind='bar', color='orange')
plt.title("Profit by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Profit")
plt.show()
```

### Analyze regional sales performance
Determine which geographical regions generate the most sales.

```python
region_sales = (df.groupby('region')['sales'].sum().sort_values(ascending=False))

print(region_sales)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(x=region_sales.index, y=region_sales.values)
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

### Analyze regional profitability:

```python
region_profit = (df.groupby('region')['profit'].sum())

print(region_profit)
```

### Analyze customer segments
Determine how different customer segments contribute to revenue.

```python
segment_sales = (df.groupby('segment')['sales'].sum().sort_values(ascending=False))

print(segment_sales)
```

```python
plt.figure(figsize=(9, 5))

segment_sales.plot(kind='bar', color='purple')
plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")

plt.show()
```

### Compare profitability:

```python
segment_profit = (df.groupby('segment')['profit'].sum())

print(segment_profit)
```

### Discount sensitivity analysis
Investigate the relationship between discount and profit

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='discount', y='profit', alpha=0.6)
plt.axhline(y=0, color='red', linestyle='--')
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

### Calculate correlation

```python
correlation = df[['discount', 'profit']].corr()

print(correlation)
```

```python
discount_profit_corr = df['discount'].corr(df['profit'])

print("Discount-Profit Correlation:", round(discount_profit_corr, 3))
```

### Group discounts into categories

```python
df['discount_group'] = pd.cut(df['discount'], bins=[-0.001, 0.1, 0.2, 0.3, 0.5, 1], labels=['0-10%', '10-20%', '20-30%', '30-50%', '50-100%'])
```

### Calculate average profit:

```python
discount_analysis = (df.groupby('discount_group', observed=False)['profit'].mean().reset_index())

print(discount_analysis)
```

### Visualize:

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=discount_analysis, x='discount_group', y='profit')
plt.title("Average Profit by Discount Level")
plt.xlabel("Discount Group")
plt.ylabel("Average Profit")
plt.show()
```

Analysis: Examine the relationship between discount levels and profit.
Business interpretation: Excessive discounting may be associated with reduced profitability.
